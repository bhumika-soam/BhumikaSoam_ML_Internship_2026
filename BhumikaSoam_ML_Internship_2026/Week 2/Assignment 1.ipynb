```python
#importing libraries
import pandas as pd
import numpy as np

#ml algos
from sklearn.tree import DecisionTreeClassifier
from sklearn.neighbors import KNeighborsClassifier
from sklearn.linear_model import LinearRegression

#splitting dataset
from sklearn.model_selection import train_test_split

#evaluate classification models
from sklearn.metrics import accuracy_score, confusion_matrix

#text to number conversion for cols
from sklearn.preprocessing import LabelEncoder
```

```python
#11
le = LabelEncoder()
print(df['Gender'].unique())
df['Gender'] = le.fit_transform(df['Gender'])

print(df['SubscriptionType'].unique())
df['SubscriptionType'] = le.fit_transform(df['SubscriptionType'])

print(df['FavoriteGenre'].unique())
df['FavoriteGenre'] = le.fit_transform(df['FavoriteGenre'])
```

```python
#12
df['SubscriptionRenewed'] = df['SubscriptionRenewed'].map({"Yes" : 1, "No" : 0})
x = df.drop(['UserID','SubscriptionRenewed'], axis = 1)
y = df['SubscriptionRenewed']
print(y)
```

```python
#13
x_train, x_test, y_train, y_test = train_test_split(
    x,y,test_size = 0.3,random_state = 95)
```

```python
#14
dt = DecisionTreeClassifier(random_state = 95)
#train
dt.fit(x_train,y_train)
predict = dt.predict(x_test)
```

```python
#15
acc = accuracy_score(y_test,predict)
print("Accuracy score : ",acc)
```

```python
#16
print("Confusion Matrix : ")
print(confusion_matrix(y_test,predict))
```

```python
#17
knn = KNeighborsClassifier(n_neighbors = 5)
knn.fit(x_train,y_train)
predict_knn = knn.predict(x_test)
acc_knn = accuracy_score(y_test,predict_knn)
print("Accuracy score of a knn : ",acc_knn)
```

```python
#18
#comparision b/n the accuracies
print("Decision tree : ",acc)
print("KNN : ", acc_knn)
if ( acc > acc_knn):
    print("Decision tree model performed better!")
elif (acc < acc_knn):
    print("KNN model performed better!")
else:
    print("Both models performed the same!")
```
